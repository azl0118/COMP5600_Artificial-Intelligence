```python
# Adam Lee
# COMP5600, Artificial Intelligence
# 8/28/26
#
# Assignment 0, Python Fundamentals
#
# Problem 1, part A
# List of model accuracies
values = [0.72, 0.81, 0.79, 0.88, 0.84]

# Print requested information
print("First accuracy:", values[0])
print("Last accuracy:", values[-1])
print("Number of values:", len(values))
print("Largest accuracy:", max(values))
print("Average accuracy:", sum(values) / len(values))

# Dictionary containing model information
modelInfo = {
    "modelName": "Baseline Classifier",
    "accuracy": 0.84,
    "numFeatures": 12
}

# Print each value from the dictionary
print("Model name:", modelInfo["modelName"])
print("Accuracy:", modelInfo["accuracy"])
print("Number of features:", modelInfo["numFeatures"])
```

```python
# part B,
# Returns "high" if the score is greater than or equal to 0.85
# Returns "medium" if the score is greater than or equal to 0.75 but less
# than 0.85
# Returns "low" otherwise
# Call the function on every value in the accuracy list and print the results

def classify_score(score):
    if score >= 0.85:
        return "high"
    elif score >= 0.75:
        return "medium"
    else:
        return "low"


for accuracy in values:
    print(accuracy, classify_score(accuracy))
```

```python
# Problem 2
#

import numpy as np

# Set the random seed
np.random.seed(42)

# Create an array of 5 random values between 0 and 1
randomValues = np.random.rand(5)

print("Array:", randomValues)
print("Shape:", randomValues.shape)

print("\nMean:", np.mean(randomValues))
print("Standard deviation:", np.std(randomValues))
print("Minimum:", np.min(randomValues))
print("Maximum:", np.max(randomValues))

# Reshape array in 5 cols, 1 row
reshapedValues = randomValues.reshape(5, 1)

print("\nReshaped array:")
print(reshapedValues)

print("\nNew shape:", reshapedValues.shape)

# Taking Array and adding 5 to every element
plus_five = reshapedValues + 5

print("\nArray after adding 5:")
print(plus_five)

# Compute dot product
dot_product = np.dot(randomValues, randomValues)

print("\nDot product:", dot_product)

# Boolean index greater than 0.5
greater_than_half = randomValues[randomValues > 0.5]

print("Values greater than 0.5:", greater_than_half)
```

Difference between the two shapes

The shape (5,) is a one-dimensional array with five values, while (5, 1) is a two-dimensional array with five rows and one column. They contain the same number of values, but the values are arranged differently.

```python
# Problem 3
import numpy as np
import matplotlib.pyplot as plt

# Create x values from 0 to 10 with an increment of 0.1
x = np.arange(0, 10.1, 0.1)

# Calculate sine and cosine
y1 = np.sin(x)
y2 = np.cos(x)

# Plot both functions
plt.plot(x, y1, linestyle='-', label='Sine')
plt.plot(x, y2, linestyle='--', label='Cosine')

# Add labels, title, legend, and gridlines
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sine and Cosine Functions')
plt.legend()
plt.grid(True)

# Save the figure
plt.savefig('trig_functions.png')

# Display the figure
plt.show()
```

One similarity and one difference between the sine and cosine
curves

Both the sine and cosine curves have a similar wave pattern and stay between -1 and 1. However, they reach their highest and lowest points at different x-values. At x = 0, cosine begins at 1, while sine begins at 0.

```python
# Problem 4
import numpy as np
import matplotlib.pyplot as plt

# Create x values from 0 through 100
x = np.arange(0, 101, 1)

# Calculate y values
y = 0.5 * x**2 + 2 * x + 1

# Print the first 5 values
print("First 5 x values:", x[:5])
print("First 5 y values:", y[:5])

# Print the shapes
print("Shape of x:", x.shape)
print("Shape of y:", y.shape)

# Find y when x = 50
y_50 = y[x == 50][0]
print("y when x = 50:", y_50)

# Find the mean of y
mean_y = np.mean(y)
print("Mean of y:", mean_y)

# Plot the quadratic function
plt.plot(x, y, linestyle='-', label='Quadratic Function')

# Mark the point where x = 50
plt.scatter(50, y_50, label='x = 50')

# Add labels, title, legend, and grid
plt.xlabel('x')
plt.ylabel('y')
plt.title('Quadratic Function')
plt.legend()
plt.grid(True)

# Save the figure
plt.savefig('quadratic_function.pdf')

# Display the figure
plt.show()
```

How does the value of (y) change as (x) becomes larger? Based on the
equation and your graph, why does the graph have this shape?

As x increases, y also increases and begins to grow more quickly. The graph bends upward because the equation includes an x squared term, which causes y to increase faster as x gets larger.
